```python
import pyobservablejs as obs

rows = [
    {"letter": "A", "frequency": 0.0812},
    {"letter": "B", "frequency": 0.0149},
    {"letter": "C", "frequency": 0.0271},
    {"letter": "D", "frequency": 0.0432},
    {"letter": "E", "frequency": 0.1202},
]
frequency_floor = 0.04

# rows and frequencyFloor cross to OJS through variables.
obs.Notebook(
    obs.md("# Python rows drive an Observable Plot bar chart"),
    obs.ojs("""
Plot.plot({
  height: 260,
  y: {grid: true},
  color: {legend: true},
  marks: [
    Plot.ruleY([frequencyFloor]),
    Plot.barY(rows, {
      x: "letter",
      y: "frequency",
      fill: (d) => d.frequency >= frequencyFloor ? "above floor" : "below floor",
      tip: true
    })
  ]
})
"""),
    variables={"rows": rows, "frequencyFloor": frequency_floor},
)
```
